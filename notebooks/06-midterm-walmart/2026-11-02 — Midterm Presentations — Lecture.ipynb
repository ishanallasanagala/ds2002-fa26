## Midterm presentations

Presentations today. Six minutes per team plus two minutes of questions, and the timing is enforced — at six minutes you stop, wherever you are. Practice it once before class; teams that have not rehearsed reliably spend four minutes on cleaning and thirty seconds on their findings.

**Structure that fits in six minutes:**

| Section | Time | What it has to do |
|---|---|---|
| The mess | 30s | Two or three specific problems, with counts |
| The pipeline | 1m | How you cleaned and joined, and one decision you had to make |
| Findings | 3m | Q1–Q5, best chart each, numbers stated |
| Recommendation | 1m | What you would actually tell them to do |
| What broke | 30s | One thing that went wrong and what you learned |

The most common failure is spending the findings time re-explaining the cleaning. The cleaning is a means; the findings are the point.

### What earns marks in the room

- **Numbers, not adjectives.** "Sales tripled" beats "sales increased significantly."
- **Stated windows and caveats.** Saying "3.2x against a 14-day baseline, and our correlation is over 14 days not 3" reads as competence, not weakness.
- **A recommendation someone could act on.** "Stock more" is not a recommendation. "Move 40% of poncho inventory to the three stores nearest the path 72 hours out" is.
- **Charts with message titles.** Week 12's rubric is already being applied here.

### What loses marks

- A correlation reported without its sample size
- A number that appears in the talk but not in the notebook
- Charts nobody can read from the back of the room
- "We ran out of time" as an answer to a question about a required section
