### What you actually learned

Not the tool list. The tool list is real — SQL, Pandas, JSON, APIs, ETL, cleaning, visualization, Git — but tools change and you can look up syntax.

What transfers is smaller and harder to look up:

- **A number is a claim, and claims need a stated basis.** "Sales tripled" needs a baseline window. A correlation needs its n. A demand figure needs to say whether the product sold out.
- **Cleaning decisions change answers.** You saw this in week one with the ponchos and again on both projects. Every decision gets recorded with the count of rows it touched.
- **Validate at the boundaries.** Row counts after every join, assertions after every pipeline stage. Most wrong answers in real work do not raise an exception.
- **Failures must not look like data.** An empty frame from a failed API call is the most dangerous thing in a pipeline.

Those four are why this course exists. The pandas syntax is how you practiced them.

### The checklist worth keeping

Paste this into the top of your next analysis, whatever the course or the job.

**Extract**
- Profile before you clean: shape, dtypes, nulls, duplicates, distinct counts
- Never mutate the source; derive from it
- Fail loudly, and label fallback data as fallback

**Transform**
- Duplicates, then types, then missing and out-of-range, then text, then dates
- Every decision logged with a row count
- Every join checked: row count before and after, `validate=`, `indicator=True`

**Analyze and communicate**
- Assert what must be true before you believe a result
- State the window, the n, and the caveat
- Titles state the finding; axes name the unit
- Restart and Run All before anyone else sees it

### One function to take with you

This is the profiling function from both project kickoffs. It is thirty seconds of work and it has caught something in every dataset in this course.

```python
import pandas as pd

def profile(df, name='dataset'):
    """First thing to run on any table you have not seen before."""
    print(f'--- {name}: {len(df):,} rows x {len(df.columns)} columns ---')
    summary = pd.DataFrame({
        'dtype': df.dtypes.astype(str),
        'nulls': df.isnull().sum(),
        'null_pct': (100 * df.isnull().mean()).round(1),
        'distinct': df.nunique(),
    })
    print(summary)
    print('duplicate rows:', df.duplicated().sum())
    return summary

demo = pd.DataFrame({
    'sku': ['PT-12', 'PT-12', 'WATER-24'],
    'qty': [2, 2, None],
    'price': ['$4.11', '$4.11', '5.00'],
})
profile(demo, 'a table you have never seen')
```

Three rows and it already tells you there is a duplicate, a missing quantity, and a price column that is text. That is the habit — look first, then work.
