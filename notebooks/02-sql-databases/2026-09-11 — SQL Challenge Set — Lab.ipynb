### Q1 — Every track with its artist's name and country.

*Expected: 9 rows, one per track.*

```python
# Selected relevant columns from both tables
q1 = q('''
SELECT tracks.title, artists.name, artists.country
FROM tracks
JOIN artists
ON tracks.artist_id = artists.artist_id
''')
q1
```

### Q2 — Which genre has the longest average track length?

Return the genre and the average, not just the name.

```python
#I selected the genre of each track and their average length and sorted by highest to lowest length
#I selected only the first row with the longest genre on average
q2 = q('''
SELECT genre, AVG(seconds) AS avg_length
FROM tracks
GROUP BY genre
ORDER BY AVG(seconds) DESC
LIMIT 1
''')
q2
```

### Q3 — For each user: how many plays, and how many distinct tracks?

Someone who played one track four times is a different listener from someone who played four different tracks. Your result should make that visible.

```python
# Grouped by user to track total play volume versus unique songs listened to
q3 = q('''
SELECT user, COUNT(*) AS plays, COUNT(DISTINCT track_id) AS distinct_tracks
FROM plays
GROUP BY user
ORDER BY plays DESC
''')
q3
```

### Q4 — Which tracks have never been played?

*Expected: 2 rows.* Hint: `LEFT JOIN` and then keep the rows where the right side came back `NULL`.

```python
# Used a LEFT JOIN to keep all tracks, then filtered for records with no matching play
q4 = q('''
SELECT tracks.track_id, tracks.title
FROM tracks
LEFT JOIN plays ON tracks.track_id = plays.track_id
WHERE plays.play_id IS NULL
''')
q4
```

### Q5 — Rank artists by total listening time.

Sum the seconds actually listened across all plays, most to least, and include a minutes column rounded to one decimal.

```python
# Joined plays to tracks and artists to calculate total seconds and minutes streamed per artist
q5 = q('''
SELECT artists.name,
       SUM(tracks.seconds) AS total_seconds,
       ROUND(SUM(tracks.seconds) / 60.0, 1) AS total_minutes
FROM plays
JOIN tracks ON plays.track_id = tracks.track_id
JOIN artists ON tracks.artist_id = artists.artist_id
GROUP BY artists.name
ORDER BY total_seconds DESC
''')
q5
```

### Q6 — Which tracks are missing a genre?

Return the track id and title. Then, in a comment, say what `WHERE genre != 'Pop'` would have done to these rows and why.

```python
# In SQL, NULL represents an unknown value rather than empty text.
# Using WHERE genre != 'Pop' would evaluate to including the tracks with a genre other than 'Pop' instead of just the tracks that are missing a genre'
#In this scenario, it would also discard the row with the actual Genre being missing and equal to Null because the Boolean value would come to unknown
q6 =q('''
SELECT track_id, title
FROM tracks
WHERE genre = NULL
''')
q6
```

### Q7 — Plays per day.

`played_on` is stored as text like `'2026-09-01'`. Count plays per date, earliest first, and include the number of distinct users active that day.

```python
q('''
SELECT played_on, COUNT(*) AS plays, COUNT(DISTINCT user) AS distinct_users
FROM plays
GROUP BY played_on
ORDER BY played_on ASC
''')
```
